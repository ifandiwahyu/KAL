1. Carilah matriks yang menggambarkan refleksi terhadap sumbu y=x. Kalikan matrik dengan 4 titik koordinat dan plot hasilnya



```python
import numpy as np
import matplotlib.pyplot as plt

# Titik-titik koordinat
koordinats = np.array([[1, 3, 2, 4], [2, 1, 3, 4]])
x = koordinats[0, :]
y = koordinats[1, :]

# Matriks refleksi
R = np.array([[0, 1],
              [1, 0]])

# Transformasi
R_trans = R @ koordinats
x_RT = R_trans[0, :]
y_RT = R_trans[1, :]

# Membuat gambar dan sumbu sumbu
fig, ax = plt.subplots()

# Plot titik-titik asal dan hasil transformasi
ax.plot(x, y, 'ro', label='Asal (Domain)')
ax.plot(x_RT, y_RT, 'bo', label='Bayangan (Codomain)')

# Hubungkan titik asal dan bayangan
for i in range(koordinats.shape[1]):
    ax.plot([x[i], x_RT[i]], [y[i], y_RT[i]], 'gray', ls='--')

# Gambar garis y = x
x_line = np.linspace(-2, 2, 100)
ax.plot(x_line, x_line, 'g--', label='Garis y = x')

# Gambar sumbu koordinat
ax.axvline(x=0, color="k", ls=":")
ax.axhline(y=0, color="k", ls=":")
ax.grid(True)
ax.axis('equal')
ax.set_xlim([-2, 2])
ax.set_ylim([-2, 2])
ax.set_title("Refleksi terhadap garis y = x")
ax.legend()
plt.show()
```

2. Carilah matriks yang menggambarkan refleksi terhadap sumbu y= -x. Kalikan matrik dengan 4 titik koordinat dan plot hasilnya



```python
import numpy as np
import matplotlib.pyplot as plt

# Titik-titik koordinat
points = np.array([[1, 2], [3, 1], [2, 3], [4, 4]])

# Matriks refleksi terhadap x = y
reflection_matrix_x_equals_y = np.array([[0, 1], [1, 0]])

# Refleksi terhadap x = y
reflected_points_x_equals_y = points @ reflection_matrix_x_equals_y

# Matriks refleksi terhadap y = -x
reflection_matrix_y_equals_neg_x = np.array([[0, -1], [-1, 0]])

# Refleksi hasil x = y terhadap y = -x
final_reflected_points = reflected_points_x_equals_y @ reflection_matrix_y_equals_neg_x

# Plotting (opsional, untuk visualisasi)
plt.plot(points[:, 0], points[:, 1], 'bo', label='Titik Asli')
plt.plot(reflected_points_x_equals_y[:, 0], reflected_points_x_equals_y[:, 1], 'go', label='Refleksi terhadap x = y')
plt.plot(final_reflected_points[:, 0], final_reflected_points[:, 1], 'ro', label='Refleksi Gabungan')

# Menambahkan garis cermin x = y
plt.plot([-5, 5], [-5, 5], 'k--', label='Garis x = y')

# Menambahkan garis cermin y = -x
plt.plot([-5, 5], [5, -5], 'k-.', label='Garis y = -x')

# Menambahkan garis hasil refleksi (opsional, untuk visualisasi)
for i in range(points.shape[0]):
    plt.plot([points[i, 0], final_reflected_points[i, 0]],
             [points[i, 1], final_reflected_points[i, 1]],
             'gray', ls='--')

plt.title('Refleksi Ganda')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.xlim([-5, 5]) # Mengatur batas sumbu x
plt.ylim([-5, 5]) # Mengatur batas sumbu y
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Definisi matriks transformasi
T1 = np.array([
    [1, 0],
    [0, -1]
])  # Refleksi terhadap sumbu x

T2 = np.array([
    [-1, 0],
    [0, 1]
])  # Refleksi terhadap sumbu y

T3 = np.array([
    [0, -1],
    [1, 0]
])  # Rotasi 90° berlawanan arah jarum jam

# Komposisi transformasi: T4 = T1 * T2 * T3
T4 = T1 @ T2 @ T3

# Vektor awal
v = np.array([2, 1])

# Hasil transformasi
result = T4 @ v

# Plot vektor awal dan hasil transformasi
plt.figure(figsize=(6, 6))
plt.plot([0, v[0]], [0, v[1]], 'b', label='Vektor Awal')
plt.plot([0, result[0]], [0, result[1]], 'r', label='Hasil Transformasi')

# Garis cermin untuk refleksi terhadap sumbu x
plt.axhline(0, color='gray', linestyle='--', label='Sumbu x')

# Garis cermin untuk refleksi terhadap sumbu y
plt.axvline(0, color='gray', linestyle='-.', label='Sumbu y')


# Tampilkan hasil
print("Hasil transformasi T4 terhadap vektor [2, 1] adalah:", result)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Transformasi Vektor')
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.show()
```
